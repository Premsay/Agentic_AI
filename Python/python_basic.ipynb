# Data Types
### Python has the following data types built-in by default, in these categories:
| Type Category | Python Types |
| :--- | :--- |
| Text | `str` |
| Numeric | `int`, `float`, `complex` |
| Sequence | `list`, `tuple`, `range` |
| Mapping | `dict` |
| Set | `set`, `frozenset` |
| Boolean | `bool` |
| Binary | `bytes`, `bytearray`, `memoryview` |
| None | `NoneType` |

```python
a = "Hello World"                # str
b = 20                           # int
c = 20.5                         # float
d = 1j                           # complex
e = [1,3,5,"hi"]                 # list      it is mutable
f = ( 1,2,"hii")                 # tuple      it is immutable
g = range(5)                     # range
h = {"name" : "John", "age" : 36}# dict
i = {"hi", "hello"}              # set         mutable
j = ({'hi', "hello"})            # frozenset   immutable you cannot add or remove elements.
k = True                         # bool
l = b"Hello"                     # bytes       immutable
m = bytearray(5)                 # bytearray   mutable
n = memoryview(bytes(5))         # memoryview
o = None                         # NoneType
```

## Strings

```python
#multiline string to a variable by using three quotes:
a = """Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua."""
print(a)
```

```python
# Strings are Arrays
a = "Hello, World!"
print(a[-10])

# String Length
print(len(a))

#To check if a certain phrase or character is present in a string, we can use the keyword in.
txt = "The best things in life are free!"
print("free" in txt)

#Print only if "free" is present:
if "free" in txt:
  print("Yes, 'free' is present.")

#Check if "expensive" is NOT present in the following text:
print("expensive" not in txt)

#print only if "expensive" is NOT present:
if "expensive" not in txt:
  print("No, 'expensive' is NOT present.")
```

## Slicing Strings

```python
#Get the characters from position 2 to position 5 (not included)
b = "Hello, World!"
print(b[2:5])

#Slice From the Start from the start to position 5
print(b[:5])

#Slice To the End  from position 2
print(b[2:])

#Negative Indexing
print(b[-5:-2])

```

## Modify Strings

```python
a = " Hello World! "
#Upper Case
print(a.upper())

#Lower Case
print(a.lower())

#Remove Whitespace The strip() method removes any whitespace from the beginning or the end:
print(a.strip()) 

#The replace() method replaces a string with another string:
print(a.replace("l", "J"))

#The split() method returns a list where the text between the specified separator becomes the list items.
name = "Prem Gupta"
first_name, last_name = name.split()
print(first_name)
print(last_name)

#Split a sentence
sentence = "I am learning Python"
words = sentence.split()
print(words)
```

## String Concatenation

```python
#Merge variable a with variable b into variable c:
a = "Hello"
b = "World"
c = a + b
print(c)

#To add a space between them, add a " "
a = "Hello"
b = "World"
c = a + " " + b
print(c)
```

## Format String / F-string

```python
'''
age = 36
This will produce an error:
txt = "My name is John, I am " + age
print(txt)
'''
#But we can combine strings and numbers by using f-strings or the format() method!
age = 36
txt = f"My name is John, I am {age}"
print(txt)

#You can also put calculations inside {}
a = 10
b = 20
print(f"Sum = {a + b}")

#Using split() and f-string together
full_name = "Prem Gupta"
first_name = full_name.split()[0]
print(f"Hello, {first_name}!")

#Number formatting
price = 99.56789
print(f"{price:.2f}") #.2f means show 2 digits after the decimal.

#Add commas to large numbers
number = 1000000
print(f"{number:,}")

#Convert to percentage
value = 0.856
print(f"{value:.2%}")

'''f"{variable}" → converts the value into a string representation.'''
```

| Escape Character | Meaning           | Example                          |
| :--------------- | :---------------- | :------------------------------- |
| `\'`             | Single Quote      | `"It\'s Python"` → `It's Python` |
| `\\`             | Backslash         | `"C:\\Users"` → `C:\Users`       |
| `\n`             | New Line          | `"Hello\nWorld"`                 |
| `\r`             | Carriage Return   | `"Hello\rHi"`                    |
| `\t`             | Tab               | `"Hello\tWorld"`                 |
| `\b`             | Backspace         | `"Hello\b"`                      |
| `\f`             | Form Feed         | `"Hello\fWorld"`                 |
| `\ooo`           | Octal value       | `"\101"` → `A`                   |
| `\xhh`           | Hexadecimal value | `"\x41"` → `A`                   |


# String Method

```python
# Case Conversion
text = "hello world"

text.capitalize()  # Hello world
text.lower()       # hello world
text.upper()       # HELLO WORLD
text.title()       # Hello World
text.swapcase()    # HELLO WORLD
text.casefold()    # hello world
```

```python
# Searching and Counting
text = "hello world hello"

text.count("hello")   # 2
text.find("world")    # 6
text.index("world")   # 6
text.rfind("hello")   # 12
text.rindex("hello")  # 12
```
